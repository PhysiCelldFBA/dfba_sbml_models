```python
import re
import sys
import json
import numpy as np
from cobra.io import validate_sbml_model
from cobra.io import read_sbml_model
from cobra.io import write_sbml_model
from cobra.manipulation.delete import remove_genes

from cobra.flux_analysis import flux_variability_analysis

sys.path.append("modules/")

from utils import correct_ids, trimm_model
```

```python
original_model_path = "models/iCB925.xml"
# output_model_path   = "models/M_barkeri_model_iMG746.xml"
final_model_path   = "models/C_beijerinckii_model_iCB925_trimmed.xml"

cb_model = read_sbml_model(original_model_path)
cb_model = correct_ids(cb_model, "ex_", "EX_")

solution = cb_model.optimize()
summary = cb_model.summary()
summary
```

```python
print(f"Trimming model {cb_model.id}")
trimmed_model = trimm_model(cb_model)
solution_trimmed = trimmed_model.optimize()
summary_trimmed = trimmed_model.summary()
print("Testing trimmed model")
assert abs(solution.objective_value-solution_trimmed.objective_value) < 1e-9
print("Success!")
print(f"Writing trimmed model to: {final_model_path}")
write_sbml_model(trimmed_model, final_model_path)
summary_trimmed
```
